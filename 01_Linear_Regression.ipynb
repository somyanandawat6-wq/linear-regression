 ### 📈 Linear Regression – Concept

**Let's understand what these 2 terms mean independently:**

- **Linear:** Progressing from one stage to another in a single series of steps (sequential), extending along a straight or nearly straight line.
- **Regression:** The terminology for regression is to determine the relationship between one **dependent variable** and a series of **independent variables**.

**💡 Thus, *Linear Regression* means:**

A linear approach for modeling the relationship between **dependent and **independent quantitative variables**. It will plot a straight line known as a **best-fit line** along the points to predict the target value.


### 🎯 Purpose of Linear Regression

- ✔ The purpose of linear regression is to **predict or forecast value** based on input.
- ✔ In linear regression, we have **input and output**. Input is termed as **independent variable** and output is **dependent variable**.
- ✔ **Example:** In house price prediction, we have two features (variables) — **number of rooms** and **price**. One is dependent variable and another is independent variable, based on the problem definition.

### 📐 Types of Linear Regression

- The linear regression line has an equation: **y = β0 + β1x (Y = mX + c)**
  - Where **Y** is the dependent variable, **X** is the independent variable, **β1 (m)** is slope and **β0 (c)** is intercept (the value of *y* when *x = 0*).

**There are 2 types of Linear Regression models:**

- **Simple Linear Regression:** A linear regression model with **one independent variable** and one dependent variable → `y = β0 + β1x1`

- **Multiple Linear Regression:** A linear regression model with **more than one independent variable** and one dependent variable → `y = β0 + β1x1 + β2x2 + ... + βnxn`

 📏 HOW TO DECIDE WHICH LINE IS BEST?

**Let us compare two lines** using the **Sum of Squared Differences** (vertical distances from data points to the line):

**Line 1 (slanted black line):**
- (2 − 1)² + (4 − 2)² + (1.5 − 3)² + (3.2 − 4)² = **7.89**

**Line 2 (horizontal blue line at y = 2.5**
- (2 − 2.5)² + (4 − 2.5)² + (1.5 − 2.5)² + (3.2 − 2.5)² = **3.99** ✅

**📊 Graph shown:**
- Data points (pink diamonds): **(1,2)**, **(2,4)**, **(3,1.5)**, **(4,3.2)**
- Two candidate lines with vertical error bars drawn to each point

**💡 Key takeaway:** The **smaller the sum of squared differences, the better the fit** of the line to the data — this is the foundation of the *Least Squares* method.

### 📰 Simple Linear Regression Model: Predicting Sunday Newspaper Circulation

**Based on weekly newspaper circulation** 📈➕📉

- Design and implement simple linear regression model to predict **Sunday newspaper circulation** based on **daily newspaper circulation**
- Predict Sunday paper circulation for **200, 300 and 500** daily circulation
- Evaluate model with **R** and **Adjusted R-square**

**Import Data Set** 


```python
import pandas as pd
df_news = pd.read_csv("NewspaperData.csv")
```

```python
df_news.head()
```

```python
df_news.info()
```

```python
import matplotlib.pyplot as plt
# Extracting daily and Sunday circulation data
daily_data = df_news['daily']
sunday_data = df_news['sunday']
# Plotting scater plot
plt.figure(figsize=(8,6))
plt.scatter(daily_data , sunday_data , color = 'blue' , alpha = 0.5)
plt.title('Scatter Plot of Daily vs Sunday Circulation')
plt.xlabel('Daily Circulation')
plt.ylabel('Sunday Circulation')
plt.show()
```

# Fitting a Linear Regression Model

```python
import statsmodels.formula.api as smf
model = smf.ols("sunday~daily", data = df_news).fit() # ordinary least square method for calc co-effi
```

```python
# regression equation has built successfully
```

```python
model.params
```

```python
#t and p-Values

print(f'model t-Values:\n{model.tvalues}\n------------\nmodel p-values: \n{model.pvalues}')
```

- **To summarise, if p value for any feature (variable) is < 0.05**, that variable is **important**, and suggest **not drop from dataset**.

```python
#Coefficients
model.params
```

```python
# to predict for 200 daily circulations
(1.339715 * 200) + 13.835630
```

```python
#R squared values
(model.rsquared,model.rsquared_adj)
```

**R-squared**

R² tells us how well the model fits the data. Closer to 1 = better fit.

Example: If R² = 0.91, the model explains 91% of the variation in y."

But R² has a quirk—it always increases when we add more features, even useless ones!

**Adjusted R-squared deals with this issue.**

Adjusted R² fixes this quirk by penalizing extra features that don't help.

Rule of thumb: If a feature doesn't improve the model enough,ed R² decreases.

Watch how Adjusted R² changes when we add/remove features later!

**Predict for new data point**


```python
# Predict for 200 and 300 daily circulations
newdata = pd.Series([200,300,500]) 
```

```python
data_pred = pd.DataFrame(newdata , columns = ['daily'])
```

```python
data_pred
```

```python
model.predict(data_pred)
```

```python
model.rsquared #92% is good score of accuracy
```

### 📈 Predicting Sales Based on Store Footfalls using Simple Linear Regression

We want to predict store sales based on footfall using a simple linear regression model.

Just like more people in a store usually means more sales, we'll use this pattern to predict future sales.

---

### Linear Regression Overview

Linear Regression is a supervised learning algorithm used to model the relationship between one or more independent variables (features) and a continuous dependent variable (target).

**Types of Linear Regression:**

**Simple (or Single) Linear Regression:**

Uses one independent variable to predict a continuous target variable by fitting a straight line


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

# Load and Preview the dataset

```python
data = pd.read_csv("General_Stores.csv")
print("Dataset Preview: ")
data.head()
```

# Define independent (Footfalls) and dependent (DailySales) variable

```python
X = data[['Footfalls']]
Y = data['DailySales']
```

X (Features/Inputs)

Double brackets ([['Footfalls']]) are used to select a DataFrame with two dimensions (rows and columns).

Scikit-learn expects the input features (X) to be in a 2D-like structure, even if there is only one feature.

Y (Target/Output)

Single brackets (['DailySales']) are used to select a Series, which is a 1D structure (just one column)

The target variable (Y) should be a 1D array or Series.


```python
type(X)
```

```python
type(Y)
```

# Visualise data and check relation between X and Y

```python
#%matplotlib inline
plt.xlabel('Footfalls')
plt.ylabel('Daily Sales')
plt.scatter(X,Y, color = 'red' , marker = 'o')
```

```python
from sklearn.linear_model import LinearRegression
```

This line imports the LinearRegression class from scikit-learn, a popular machine learning library.

It allows us to create a simple linear regression model,

which finds the best-fit straight line to predict a target variable based on input.

**Create and train the linear regression model**


```python
model = LinearRegression()
model.fit(X , Y)
```

```python
#fit is a method for creating model or training data or
#creating relations between x and y
```

# Print the model Parameters

```python
print(f"Coefficient (Slope): {model.coef_}") #  β1
print(f"Intercept: {model.intercept_}") #   β0
```

# simple linear regresson equation y = β0 + β1*x


Where:

y = Predicted value (dependent variable, e.g., sales)

x = Input value (independent variable, e.g., footfall)

β1 = Slope (coefficient showing how much y changes with x)

β0= Intercept (value of y when x=0 )


# (1) Predict sale of with footfall of 20

```python
#y = β0 + β1*x
model.intercept_ + model.coef_ * 20
```

```python
model.predict([[20]])
```

```python
# Quiz Can you calculate and confirm what will be
#prediction value for footfall of 35
```

```python
#y = β0 + β1*x
model.intercept_ + model.coef_ * 35
```

```python
model.predict([[35]])
```

# Predict sales using the trained model for new footfall values


```python
# Prepare new footfall values (for example: 45, 50, 60)
new_footfall = pd.DataFrame({'Footfalls' : [45, 50, 60]})
new_footfall
```

```python
# Predict new sales
predicted_sales = model.predict(new_footfall)
print("Predicted Daily Sales for new Footfalls:\n\n")
predicted_sales
```

```python
print(pd.DataFrame({'Footfalls' : new_footfall['Footfalls'] , 'Predicted Sales' : predicted_sales }))
```
